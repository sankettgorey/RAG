```python
files = ['./Data/-0___jonew__judis__6080.pdf', './Data/-0___jonew__judis__5201.pdf', './Data/-0___jonew__judis__4924.pdf']

for file in files:
    loader = PyPDFLoader(file)

    text = loader.load()



    # Step 2: Sentence-aware split
    sentence_splitter = SpacyTextSplitter(
        chunk_size=2000,      # larger because sentences are preserved
        chunk_overlap=200
    )
    sentence_chunks = sentence_splitter.split_documents(text)

    # Step 3: Apply recursive fallback if sentence chunks are still too large
    final_chunks = []
    recursive_splitter = RecursiveCharacterTextSplitter(
        chunk_size=1000,
        chunk_overlap=200
    )

    for chunk in sentence_chunks:
        if len(chunk.page_content) > 1200:  # arbitrary threshold
            smaller_chunks = recursive_splitter.split_documents([chunk])
            final_chunks.extend(smaller_chunks)
        else:
            final_chunks.append(chunk)

    print(f"Original pages: {len(text)}")
    print(f"Sentence chunks: {len(sentence_chunks)}")
    print(f"Final chunks after hybrid split: {len(final_chunks)}")



    embedding_function = OllamaEmbeddings(model = 'nomic-embed-text:latest')

    vectorstore.add_documents(documents = final_chunks)
```

```python
from langchain_community.vectorstores import Chroma
from langchain_ollama import OllamaEmbeddings, ChatOllama

from langchain.chains import RetrievalQA
```

```python
embedding_function = OllamaEmbeddings(model = 'nomic-embed-text:latest')


try:
    vectorstore = Chroma.from_documents(
        documents = final_chunks,
        embedding = embedding_function,
        persist_directory = './chroma_db1',
        collection_name='legal_chunks'
    )
except Exception as e:
    print(str(e))
```

```python

query = "What is the final judgment in this case?"
results = vectorstore.similarity_search(query, k=5)

for i, doc in enumerate(results):
    print(f"\n--- Result {i+1} ---")
    print(doc.page_content)  # preview first 500 chars
```

```python
print("Loaded Chroma with", vectorstore._collection.count(), "chunks")
```

# Retrieval Pipeline

```python
retriever1 = vectorstore.as_retriever(
    search_type="similarity",   # or "mmr" (Max Marginal Relevance)
    search_kwargs={"k": 5}      # number of chunks to retrieve
)


# llm = ChatOllama(model = 'llama3.2:3b')
llm = ChatOllama(model = 'mistral:7b-instruct')
# llm = ChatOllama(model = 'qwen2.5:7b-instruct')



qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    retriever=retriever1,
    return_source_documents=True   # so you can see which chunks were used
)



query = "What is the main judgment in this case?"
result = qa_chain.invoke({"query": query})

print("\nAnswer:\n", result["result"])
print("\nSources:")
for doc in result["source_documents"]:
    print("---")
    print(doc.page_content[:300])  # preview source
```
